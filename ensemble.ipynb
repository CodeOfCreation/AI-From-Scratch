```python
import numpy as np

predictions = np.array([100, 120, 110])

# Simple averaging
final_prediction = np.mean(predictions)
print(final_prediction)  # 110.0

# Weighted averaging
weights = np.array([2, 1, 3])

weighted_prediction = np.average(
    predictions, weights=weights
)
print(weighted_prediction)  # 108.33
```

```python
from collections import Counter
import numpy as np

# Hard voting
predictions = ["spam", "spam", "not_spam"]
final_prediction = Counter(predictions).most_common(1)[0][0]

print(final_prediction)  # spam

# Soft voting: probabilities for classes [not_spam, spam]
probabilities = np.array([
    [0.20, 0.80],
    [0.40, 0.60],
    [0.70, 0.30]
])

average_probabilities = probabilities.mean(axis=0)
classes = ["not_spam", "spam"]

print(classes[np.argmax(average_probabilities)])
# spam
```
