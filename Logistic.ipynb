```python
import numpy as np

# One feature: study hours
X = np.array([[1], [2], [3], [4]])
y = np.array([0, 0, 1, 1])

w = np.zeros(1)
b = 0.0
learning_rate = 0.1

for epoch in range(20):
    errors = 0

    for xi, target in zip(X, y):
        score = np.dot(xi, w) + b
        prediction = int(score >= 0)

        # Update only if misclassified
        error = target - prediction

        if error != 0:
            w += learning_rate * error * xi
            b += learning_rate * error
            errors += 1

    if errors == 0:
        break

print("Weight:", w)
print("Bias:", b)

scores = X @ w + b
predictions = (scores >= 0).astype(int)
print("Predictions:", predictions)
```

```python
from sklearn.linear_model import Perceptron

model=Perceptron(max_iter=100,eta0=0.1,random_state=42)
model.fit(X,y)
predictions=model.predict(X)
print('weights:',model.coef_)
print('bias:',model.intercept_)
print('predictions:',predictions)
```

```python
#Sigmoid 
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

scores = np.array([-5, -2, 0, 2, 15])
probabilities = sigmoid(scores)

print(probabilities)
```

```python
X = np.array([[1.], [2.], [3.], [4.]])
y = np.array([0., 0., 1., 1.])

w = np.zeros(1)
b = 0.0
lr = 0.1

for _ in range(2000):
    p = sigmoid(X @ w + b)

    dw = X.T @ (p - y) / len(y)
    db = np.mean(p - y)

    w -= lr * dw
    b -= lr * db

print("Weight:", w)
print("Bias:", b)
print("Probabilities:", sigmoid(X @ w + b))
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import log_loss

X = [[1], [2], [3], [4]]
y = [0, 0, 1, 1]

model = LogisticRegression()
model.fit(X, y)

probs = model.predict_proba(X)[:, 1]

print("Probabilities:", probs)
print("Log loss:", log_loss(y, probs))
```
