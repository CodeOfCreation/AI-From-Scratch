```python
from torch.utils.data import TensorDataset,DataLoader

x=torch.tensor([[1],[2],[3],[4]])
y=torch.tensor([[2],[4],[6],[8]])

dataset=TensorDataset(x,y)
loader=DataLoader(dataset,batch_size=2,shuffle=True)

for input,target in loader:
    print(input,target)
```
