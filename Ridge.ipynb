```python
import numpy as np

x = np.array([
    [1., 2.],
    [2., 1.],
    [3., 4.],
    [4., 3.]
])

y = np.array([5., 4., 11., 10.])

x_mean=x.mean()
x_std=x.std()
x_scaled=(x-x_mean)/x_std

X_design = np.column_stack([
    np.ones(len(x_scaled)),
    x_scaled
])
w = np.zeros(X_design.shape[1])
learning_rate = 0.05
alpha = 0.1
epochs = 100
n = len(y)
for i in range(epochs):
  y_pred=X_design @ w
  error=y_pred-y
  grad=(2/n)*(X_design.T @ error)
  grad[1:]+=2*alpha*w[1:]
  w=w-(learning_rate*grad)

print("Intercept:", w[0])
print("Coefficients:", w[1:])
```

```python
import numpy as np

X = np.array([
    [1., 2.],
    [2., 1.],
    [3., 4.],
    [4., 3.],
    [5., 5.]
])
y = np.array([5., 5., 11., 11., 15.])

# Standardize features
mean = X.mean(axis=0)
std = X.std(axis=0)
X = (X - mean) / std

n, p = X.shape
w = np.zeros(p)
b = 0.0

learning_rate = 0.01
alpha = 0.1
epochs = 5000

for _ in range(epochs):
    y_pred = X @ w + b

    # Gradient of MSE (no L1 term here)
    dw = (-2 / n) * X.T @ (y - y_pred)
    db = (-2 / n) * np.sum(y - y_pred)

    # Gradient step for weights and bias
    w_temp = w - learning_rate * dw
    b -= learning_rate * db

    # L1 proximal step: soft-threshold weights
    threshold = learning_rate * alpha
    w = np.sign(w_temp) * np.maximum(
        np.abs(w_temp) - threshold, 0
    )

predictions = X @ w + b
loss = np.mean((y - predictions) ** 2) + alpha * np.sum(np.abs(w))

print("Weights:", w)
print("Bias:", b)
print("Lasso loss:", loss)
```

```python
# elastic net regression

import numpy as np

X = np.array([
    [1., 2.],
    [2., 1.],
    [3., 4.],
    [4., 3.],
    [5., 5.]
])
y = np.array([5., 5., 11., 11., 15.])

# Standardize features
mean = X.mean(axis=0)
std = X.std(axis=0)
X = (X - mean) / std

n, p = X.shape
w = np.zeros(p)
b = 0.0

learning_rate = 0.01
alpha = 0.1
l1_ratio = 0.5
epochs = 5000

for _ in range(epochs):
    y_pred = X @ w + b

    # Gradient of half-MSE
    dw = -(X.T @ (y - y_pred)) / n
    db = -np.mean(y - y_pred)

    # Gradient step for data loss + L2 penalty
    w_temp = w - learning_rate * (
        dw + alpha * (1 - l1_ratio) * w
    )
    b -= learning_rate * db

    # Proximal step for L1 penalty
    threshold = learning_rate * alpha * l1_ratio
    w = np.sign(w_temp) * np.maximum(
        np.abs(w_temp) - threshold, 0
    )

predictions = X @ w + b

loss = (
    0.5 * np.mean((y - predictions) ** 2)
    + alpha * l1_ratio * np.sum(np.abs(w))
    + 0.5 * alpha * (1 - l1_ratio) * np.sum(w ** 2)
)

print("Weights:", w)
print("Bias:", b)
print("Elastic Net loss:", loss)
```
