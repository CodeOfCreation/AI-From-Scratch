```python
import numpy as np

class MiniGD:
    def __init__(self,learning_rate=0.01,epochs=100,batch_size=32,random_state=42):
        self.lr=learning_rate
        self.epochs=epochs
        self.w=None
        self.b=0.0
        self.random_state=random_state
        self.batch_size=batch_size
        self.loss_history=[]

    def fit(self,x,y):
        x=np.asarray(x,dtype=float)
        y=np.asarray(y,dtype=float).reshape(-1)

        if x.ndim != 2:
            raise ValueError("X must be a 2D array.")
        if len(x) != len(y) or len(y) == 0:
            raise ValueError("X and y must have equal, nonzero lengths.")
        if not np.isfinite(x).all() or not np.isfinite(y).all():
            raise ValueError("X and y must contain only finite values.")
        if self.batch_size < 1:
            raise ValueError("batch_size must be at least 1.")

        n_samples,n_features=x.shape

        self.w=np.zeros(n_features)
        self.b=0.0
        self.loss_history=[]

        rng=np.random.default_rng(self.random_state)

        for epoch in range(self.epochs):
            indices=rng.permutation(n_samples)
            x_shuffled=x[indices]
            y_shuffled=y[indices]

            for start in range(0,n_samples,self.batch_size):
                end=start+self.batch_size
                x_batch=x_shuffled[start:end]
                y_batch=y_shuffled[start:end]
                m=len(y_batch)
                y_pred=x_batch @ self.w+self.b
                error=y_pred-y_batch
                dw=(2/m)*(x_batch.T @ error)
                db=(2/m)*np.mean((error))

                self.w=self.w-self.lr*dw
                self.b=self.b-self.lr*db


            epoc_pred=x @ self.w+self.b
            loss=np.mean((epoc_pred-y)**2)
            self.loss_history.append(loss)

        return self

    def predict(self, X):
        if self.w is None:
            raise ValueError("Fit the model before predicting.")

        X = np.asarray(X, dtype=float)

        if X.ndim != 2:
            raise ValueError("X must be a 2D array.")
        return X @ self.w + self.b
```

```python

from sklearn.preprocessing import StandardScaler
rng = np.random.default_rng(42)

X = np.arange(1, 101, dtype=float).reshape(-1, 1)
y = 4 * X[:, 0] + 3 + rng.normal(0, 2, len(X))

# Scale X to make gradient updates more stable
x_scaler = StandardScaler()
X_scaled = x_scaler.fit_transform(X)

model = MiniGD(
    learning_rate=0.01,
    epochs=100,
    batch_size=16,
    random_state=42
)

model.fit(X_scaled, y)

# Convert parameters back to the original X scale
weight = model.w[0] / x_scaler.scale_[0]
bias = model.b - weight * x_scaler.mean_[0]

print("Learned weight:", weight)
print("Learned bias:", bias)
print("Final MSE:", model.loss_history[-1])
```
