```python
import numpy as np

class LinearRegressionGD:
    def __init__(self,learning_rate=0.01,epochs=100):
        self.lr=learning_rate
        self.epochs=epochs
        self.w=0
        self.b=0
        self.loss_history=[]

    def predict(self,x):
        X = np.asarray(X)
        return X @ self.weights + self.bias

    def compute_loss(self,x,y):
        y_pred=self.predict(x)
        return np.mean((y_pred-y)**2)

    def fit(self,x,y):
        n=len(x)
        for i in range(self.epochs):
            y_pred=self.predict(x)
            error=y_pred-y
            dw=(2/n)*np.sum(x*error)
            db=(2/n)*np.sum(error)

            self.w=self.w-self.lr*dw
            self.b=self.b-self.lr*db

            loss=self.compute_loss(x,y)
            self.loss_history.append(loss)

        return self
    
```

```python
# Training data: hours studied
x = np.array([1, 2, 3, 4, 5], dtype=float)

# Target: exam marks
y = np.array([15, 25, 35, 45, 55], dtype=float)

model = LinearRegressionGD(
    learning_rate=0.01,
    epochs=2000
)

model.fit(x, y)

print("Weight:", model.w)
print("Bias:", model.b)
print("Final MSE:", model.loss_history[-1])

print("Prediction for 6 hours:",
      model.predict(np.array([6.0])))
```

```python
X = df.select_dtypes(include='number').drop(columns=['Exam_Score'])
y=df['Exam_Score']

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
model1=LinearRegression()
model1.fit(x_train,y_train)
y_pred=model1.predict(x_test)

from sklearn.metrics import r2_score,mean_squared_error

print(r2_score(y_test,y_pred))
print(mean_squared_error(y_test,y_pred))
```
