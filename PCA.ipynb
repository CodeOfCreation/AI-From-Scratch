```python
import numpy as np

X = np.array([
    [1, 10],
    [2, 20],
    [3, 30]
], dtype=float)

# 1. Center the data
X_centered = X - X.mean(axis=0)

# 2. Direction of the diagonal
pc1 = np.array([1, 10], dtype=float)
pc1 = pc1 / np.linalg.norm(pc1)

# 3. Project each point onto PC1
X_reduced = X_centered @ pc1

print(X_centered)
print(X_reduced)
```

```python
import numpy as np

X = np.array([
    [2, 1],
    [3, 2],
    [4, 3],
    [5, 4]
], dtype=float)

# Step 1: Center the data
mean = X.mean(axis=0)
X_centered = X - mean

# Step 2: Covariance matrix
C = np.cov(X_centered, rowvar=False)

# Step 3: Eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eigh(C)

# Step 4: Sort in descending order
idx = np.argsort(eigenvalues)[::-1]
eigenvalues = eigenvalues[idx]
eigenvectors = eigenvectors[:, idx]

# Step 5: Keep only PC1
W = eigenvectors[:, :1]

# Step 6: Project data
X_reduced = X_centered @ W

print("Covariance matrix:\n", C)
print("Eigenvalues:", eigenvalues)
print("PC1 direction:\n", W)
print("Reduced data:\n", X_reduced)
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=1)

X_reduced = pca.fit_transform(X)

print("Reduced data:\n", X_reduced)
print("PC1 direction:", pca.components_)
print("Explained variance ratio:",
      pca.explained_variance_ratio_)
```
