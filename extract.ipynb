```python
import numpy as np

X = np.array([
    [25, 30],
    [26, 32],
    [27, 34],
    [26, np.nan]
], dtype=float)

# Use Age to find the two nearest rows to row D
distances = np.abs(X[:-1, 0] - X[-1, 0])

nearest = np.argsort(distances)[:2]

# Estimate Salary from those neighbors
X[-1, 1] = np.mean(X[nearest, 1])

print(X)
```

```python
import numpy as np
from sklearn.experimental import enable_iterative_imputer  # noqa
from sklearn.impute import IterativeImputer
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

X = np.array([
    [1.0, 30.0],
    [2.0, 40.0],
    [3.0, 50.0],
    [4.0, np.nan]
])

imputer = IterativeImputer(
    max_iter=10,
    random_state=42
)

X_imputed = imputer.fit_transform(X)
print(X_imputed)
```
