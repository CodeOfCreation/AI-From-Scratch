```python
import numpy as np

# Two features per observation
X = np.array([
    [1, 1],
    [2, 1],
    [3, 2],
    [4, 3],
    [5, 3],
    [6, 4]
], dtype=float)

y = np.array([6, 8, 13, 17, 20, 25], dtype=float)

# Add a column of ones for the intercept
X_b = np.c_[np.ones(X.shape[0]), X]

# Learn coefficients
coefficients = np.linalg.lstsq(X_b, y, rcond=None)[0]

# Predict
y_pred = X_b @ coefficients

# Evaluate
mse = np.mean((y - y_pred) ** 2)

print("Intercept:", coefficients[0])
print("Feature coefficients:", coefficients[1:])
print("Predictions:", y_pred)
print("MSE:", mse)

# Predict for a new observation
new_X = np.array([1, 7, 4])
new_prediction = new_X @ coefficients

print("New prediction:", new_prediction)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X, y)

print("Intercept:", model.intercept_)
print("Coefficients:", model.coef_)
print("Predictions:", model.predict(X))
```

```python
import numpy as np

# Training data
X = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([3, 5, 7, 9, 11], dtype=float)

# Standardize X to help gradient descent converge
x_mean = X.mean()
x_std = X.std()
X_scaled = (X - x_mean) / x_std

# Initialize parameters
w = 0.0
b = 0.0

learning_rate = 0.1
epochs = 1000
n = len(y)

for epoch in range(epochs):
    # 1. Predict
    y_pred = w * X_scaled + b

    # 2. Calculate gradients of MSE
    dw = (2 / n) * np.sum(X_scaled * (y_pred - y))
    db = (2 / n) * np.sum(y_pred - y)

    # 3. Update parameters
    w -= learning_rate * dw
    b -= learning_rate * db

# Convert the learned slope back to the original X scale
original_w = w / x_std
original_b = b - w * x_mean / x_std

print("Slope:", original_w)
print("Intercept:", original_b)
print("Predictions:", original_w * X + original_b)
```

```python
from sklearn.linear_model import SGDRegressor
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

X = np.array([1, 2, 3, 4, 5], dtype=float).reshape(-1, 1)
y = np.array([3, 5, 7, 9, 11], dtype=float)

model = make_pipeline(
    StandardScaler(),
    SGDRegressor(
        loss="squared_error",
        learning_rate="constant",
        eta0=0.01,
        max_iter=5000,
        tol=1e-8,
        random_state=42
    )
)

model.fit(X, y)
print("Predictions:", model.predict(X))
```
