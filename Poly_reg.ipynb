```python

import numpy as np

def poly_reg(x, degree):
    x = np.asarray(x, dtype=float)

    # Convert 1D input into a 2D array
    if x.ndim == 1:
        x = x.reshape(-1, 1)

    # This implementation supports only one input feature
    if x.shape[1] != 1:
        raise ValueError("Only one input feature is supported")

    features = [x ** power for power in range(1, degree + 1)]

    # Add the intercept column of ones
    x_poly = np.column_stack([np.ones(x.shape[0])] + features)

    return x_poly


class Poly:
    def __init__(self, degree=2):
        self.degree = degree
        self.coef = None

    def fit(self, x, y):
        x_poly = poly_reg(x, self.degree)
        y = np.asarray(y, dtype=float).ravel()

        self.coef = np.linalg.lstsq(
            x_poly, y, rcond=None
        )[0]

        return self

    def predict(self, x):
        if self.coef is None:
            raise ValueError("Fit the model before predicting")

        x_poly = poly_reg(x, self.degree)

        return x_poly @ self.coef


X = np.array([1, 2, 3, 4, 5])
y = np.array([3, 6, 11, 18, 27])

model = Poly(degree=2)
model.fit(X, y)

predictions = model.predict(X)

print("Coefficients:", model.coef)
print("Actual:     ", y)
print("Predicted:  ", np.round(predictions, 4))
```
