```python
import numpy as np
from sklearn.preprocessing import FunctionTransformer

log_transformer = FunctionTransformer(
    np.log1p,
    feature_names_out="one-to-one"
)

X = np.array([[0], [9], [99]])
print(log_transformer.transform(X))
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import Pipeline

tr=FunctionTransformer(np.log1p,feature_names_out='one-to-one')
model=Pipeline([('Function',tr),
                ('regression',LinearRegression())])

x=np.array(([0],[2],[3],[6],[9]),dtype=float)
y=np.array((1,2,3,5,6),dtype=float)

model.fit(x,y)
print(model.predict([[4]]))
```

```python
x=np.array([1,2,3,4,5,16],dtype=float)
lam=0.5

if lam==0:
    x_transformed=np.log(x)
else:
    x_transformed=(x**lam-1)/lam

print(x_transformed)
```

```python
import numpy as np
from sklearn.preprocessing import PowerTransformer

X = np.array([
    [1, 10],
    [4, 20],
    [9, 30],
    [16, 40],
    [100, 50]
], dtype=float)

# Yeo-Johnson works with positive, zero and negative values
transformer = PowerTransformer(
    method="yeo-johnson",
    standardize=True
)

X_transformed = transformer.fit_transform(X)

print(X_transformed)
print("Learned lambdas:", transformer.lambdas_)
```
