```python
import numpy as np

x = np.array([1., 2., 3., 4., 5.])
y = np.array([3., 5., 7., 9., 11.])

w = 0.0
b = 0.0
learning_rate = 0.01
epochs = 100
n=len(x)
np.random.seed(42)

for epoch in range(epochs):
    indices = np.random.permutation(n)

    for i in indices:
        xi = x[i]
        yi = y[i]

        y_pred = w * xi + b

        dw = 2 * xi * (y_pred - yi)
        db = 2 * (y_pred - yi)
        w -= learning_rate * dw
        b -= learning_rate * db
    predictions = w * x + b
    loss = np.mean((predictions - y) ** 2)

print("Weight:", w)
print("Bias:", b)
print("Final MSE:", loss)
```

```python
X = np.array([1., 2., 3., 4., 5.])
y = np.array([3., 5., 7., 9., 11.])

w, b = 0.0, 0.0
learning_rate = 0.01
batch_size = 2
epochs = 100

n = len(X)
np.random.seed(42)

for epoch in range(epochs):
    indices = np.random.permutation(n)

    for start in range(0, n, batch_size):
        batch_indices = indices[start:start + batch_size]

        X_batch = X[batch_indices]
        y_batch = y[batch_indices]

        # Predictions for this batch
        y_pred = w * X_batch + b
        errors = y_pred - y_batch

        # Average gradients over the batch
        dw = (2 / len(X_batch)) * np.sum(X_batch * errors)
        db = (2 / len(X_batch)) * np.sum(errors)

        # Update once per batch
        w -= learning_rate * dw
        b -= learning_rate * db

    # Full-dataset loss for monitoring
    predictions = w * X + b
    loss = np.mean((predictions - y) ** 2)

print("Weight:", w)
print("Bias:", b)
print("Final MSE:", loss)
```
